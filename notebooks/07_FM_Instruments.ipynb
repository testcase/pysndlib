# The Classic FM Instruments

CLM's instrument collection is largely a study in FM synthesis. This notebook walks through the classics in `pysndlib.instruments`, tweaking the parameters that matter.

Source: [FM documentation -- instrument chapters](https://ccrma.stanford.edu/software/snd/snd/fm.html)

```python
import numpy as np
import matplotlib.pyplot as plt
import IPython.display as ipd

import pysndlib.clm as clm
from pysndlib.instruments import fm_bell, fm_drum, fm_trumpet, fm_violin

SR = 44100

def spectrum(x, n=4096):
    rl = np.copy(x[:n].astype(float)); im = np.zeros(n)
    clm.mus_fft(rl, im, n, 1)
    clm.mus_rectangular2polar(rl, im)
    return np.abs(rl[:n//2])
```

## fm_bell: inharmonic beauty

Bells sound inharmonic because their partials are not integer ratios. fm_bell achieves this with an inharmonic FM ratio (and an index envelope that keeps the strike bright while the body decays):

```python
with clm.Sound('bell.wav', channels=1, srate=SR):
    fm_bell(0.0, 3.0, 440.0, 0.4)
    fm_bell(1.0, 3.0, 440.0 * 1.5, 0.3)   # a "fifth" up
```

```python
ipd.Audio('bell.wav', rate=SR)
```

```python
bd, _ = clm.file2ndarray('bell.wav')
f = np.arange(2048) * SR / 4096
plt.figure(figsize=(14, 3))
plt.plot(f, spectrum(bd[0][:4096]), lw=0.8)
plt.xlim(0, 6000); plt.title('bell: inharmonic partials'); plt.show()
```

## fm_drum: sweeping the index fast

fm_drum's index envelope has a rapid sweep -- the "whoomp" of a drum comes from the sidebands spreading outward faster than the ear can track. Its `index` parameter scales the sweep:

```python
with clm.Sound('drum.wav', channels=1, srate=SR):
    fm_drum(0.0, 0.6, 80.0, 0.5, 2.0)
    fm_drum(0.8, 0.6, 80.0, 0.5, 5.0)    # bigger sweep
```

```python
ipd.Audio('drum.wav', rate=SR)
```

```python
dd, _ = clm.file2ndarray('drum.wav')
plt.figure(figsize=(14, 4))
plt.specgram(dd[0], NFFT=1024, Fs=SR, noverlap=256)
plt.ylim(0, 6000); plt.title('drum index sweeps'); plt.show()
```

## fm_trumpet: brightness follows loudness

Brass sounds start dull and get bright as they get louder -- in FM terms the index envelope follows the amplitude envelope. Listen to the attack "bloom":

```python
with clm.Sound('trumpet.wav', channels=1, srate=SR):
    fm_trumpet(0.0, 1.5, frq1=250, frq2=250)
```

```python
ipd.Audio('trumpet.wav', rate=SR)
```

```python
td, _ = clm.file2ndarray('trumpet.wav')
plt.figure(figsize=(14, 4))
plt.specgram(td[0], NFFT=2048, Fs=SR, noverlap=512)
plt.ylim(0, 6000); plt.title('trumpet: brightness blooming with the attack'); plt.show()
```

## fm_violin: vibrato and index together

The violin adds slow periodic and random vibrato on top of a moderate index:

```python
with clm.Sound('violin2.wav', channels=1, srate=SR):
    fm_violin(0.0, 2.0, 294.0, 0.4)               # D4
    fm_violin(0.5, 2.0, 294.0, 0.4, fm_index=3.0)  # same note, brighter
```

```python
ipd.Audio('violin2.wav', rate=SR)
```
